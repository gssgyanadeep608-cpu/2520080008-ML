```python
!pip install umap-learn -q

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import umap
from sklearn.ensemble import IsolationForest

# Load high-dimensional dataset (Handwritten Digits: 64 features)
digits = load_digits()
X, y = digits.data, digits.target
print("Dataset ready! Shape:", X.shape)
```

```python
# 1. PCA (Linear reduction)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# 2. t-SNE (Non-linear reduction)
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X)

# Plotting PCA vs t-SNE
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

scatter1 = axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='tab10', s=10, alpha=0.7)
axes[0].set_title('PCA Reduction')
axes[0].grid(True)

scatter2 = axes[1].scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='tab10', s=10, alpha=0.7)
axes[1].set_title('t-SNE Reduction')
axes[1].grid(True)

plt.show()
```

```python
# Initialize Isolation Forest
iso_forest = IsolationForest(contamination=0.05, random_state=42)
preds = iso_forest.fit_predict(X)

# Isolation Forest outputs 1 for inliers and -1 for anomalies
anomalies = np.where(preds == -1)[0]
print(f"Number of detected anomalies/outliers: {len(anomalies)}")
```
