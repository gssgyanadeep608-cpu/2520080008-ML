```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Load dataset
data = load_breast_cancer()
X, y = data.data, data.target
print("Dataset ready! Shape:", X.shape)
```

```python
# Initialize Stratified K-Fold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
model = RandomForestClassifier(random_state=42)

# Compute cross-validation scores
cv_scores = cross_val_score(model, X, y, cv=cv, scoring='accuracy')

print(f"Fold Accuracies: {np.round(cv_scores, 4)}")
print(f"Mean CV Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std() * 2:.4f})")
```

```python
# Inner and Outer folds for Nested CV
inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
outer_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Hyperparameter grid to search
param_grid = {
    'n_estimators': [50, 100],
    'max_depth': [3, 5, None]
}

# Base model and Grid Search wrapper (acts as the inner loop optimizer)
clf = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(estimator=clf, param_grid=param_grid, cv=inner_cv)

# Run nested cross-validation across the outer folds
nested_scores = cross_val_score(grid_search, X=X, y=y, cv=outer_cv)

print(f"Nested CV Scores across outer folds: {np.round(nested_scores, 4)}")
print(f"Mean Nested CV Accuracy: {nested_scores.mean():.4f}")
```
